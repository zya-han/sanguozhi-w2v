```python
from gensim.models import Word2Vec

model = Word2Vec.load('./models/w2v_sanguozhi.model')
```

```python
# 두 단어의 유사 단어 목록 비교
myword1 = '荀彧'
myword2 = '郭嘉'

print(model.wv.most_similar(myword1, topn=20))
print(model.wv.most_similar(myword2, topn=20))
```

```python
# 우리 중에 스파이가 있는 것 같아
mygroup = ['周瑜', '魯肅', '呂蒙', '諸葛亮']

print(model.wv.doesnt_match(mygroup))
```
